```python
from genehearing.common.tools import load_config
```

```python
config = load_config()
```

```python
match_column = config['pesel_columnname']
tonaldataname=config["tonaldataname"]
tonal_audiometry_datapath = config["datarawdirectory"] + tonaldataname + '_final'+'.csv'

encodings_to_try = ["utf-8", "utf-8-sig", "cp1250", "latin1"]
for enc in encodings_to_try:
    try:
        df = pd.read_csv(f"data/raw/{config['genetic_dataname_all']}.csv", sep=None, engine='python', dtype={match_column: str}, encoding=enc)
        df_audiometry = pd.read_csv(tonal_audiometry_datapath, sep=None, engine='python', dtype={match_column: str}, encoding=enc)
        print(f"Wczytano poprawnie z encoding='{enc}'")
    except UnicodeDecodeError:
        print(f"Nieudane wczytanie przy encoding='{enc}'")
```

```python
for value in df['PESEL'].values:
    print(value, type(value))
```

```python
for value in df_audiometry['PESEL'].values:
    print(value, type(value))
```

```python
merged = pd.merge(df_audiometry, df, how='left', on='PESEL')
merged = merged.loc[:, ~merged.columns.str.contains('^Unnamed')]
```
