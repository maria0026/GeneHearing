```python
from genehearing.common.tools import load_config
```

```python
config = load_config()
```

```python
match_column = config['pesel_columnname']
tonaldataname=config["tonaldataname"]
tonal_audiometry_datapath = config["datarawdirectory"] + tonaldataname + '_final'+'.csv'

encodings_to_try = ["utf-8", "utf-8-sig", "cp1250", "latin1"]
for enc in encodings_to_try:
    try:
        df = pd.read_csv(f"data/raw/{config['genetic_dataname_all']}.csv", sep=None, engine='python', dtype={'PESEL': str}, encoding=enc)
        df_audiometry = pd.read_csv(tonal_audiometry_datapath, sep=None, engine='python', dtype={'pesel': str}, encoding=enc)
        df_audiometry.columns = df_audiometry.columns.str.upper()
        print(f"Wczytano poprawnie z encoding='{enc}'")
    except UnicodeDecodeError:
        print(f"Nieudane wczytanie przy encoding='{enc}'")
```

```python
df_implanty = pd.read_csv(f"data/raw/implanty_final.csv", sep=None, engine='python', dtype={'PESEL': str}, encoding=enc)
df_implanty_aparaty = pd.read_csv(f"data/raw/implanty_aparaty_final.csv", sep=None, engine='python', dtype={'PESEL': str}, encoding=enc)
df_abr =  pd.read_csv(f"data/raw/ABR.csv", sep=None, engine='python', dtype={'PESEL': str}, encoding=enc)
df_implanty_geny =  pd.read_csv(f"data/raw/implanty_geny.csv", sep=None, engine='python', dtype={'PESEL': str}, encoding=enc)
```
